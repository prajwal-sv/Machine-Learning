```python
# USing OLS
import numpy as np 
# calculate mean 
def mean(arr):
    return np.sum(arr) / len(arr)
# calculate slop
def calculate_slope(x,y):
    x_mean = mean(x)
    y_mean = mean(y)

    numerator = np.sum((x-x_mean)*(y-y_mean))

    denominator  = np.sum((x-x_mean)**2)

    slop =  numerator/denominator

    return slop



# calculate intercept 
def calculate_intercept(x,y,m):
    x_mean = mean(x)
    y_mean = mean(y)
    intercept = y_mean - m * x_mean

    return intercept

# predict
def predict(x_value , m , b):
    return m*x_value+b


# mean squred error 
def calculate_mse(y_actual, y_predicted):
    return np.mean((y_actual -  y_predicted)**2)
# dataset
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 30, 40, 50])


# Training 
m = calculate_slope(x, y)
b = calculate_intercept(x, y, m)
# Prediction
prediction = predict(6, m, b)
# mse
mse = calculate_mse(y,prediction)

print("Slope (m):", m)
print("Intercept (b):", b)
print("Predictions:", prediction)
print("MSE:", mse)



```

```python
# Using Gradient Decent 
import numpy as np 
def predict(x,m,b):
    return m * x+b

def calculate_mse(y , y_pred):
    return np.mean((y-y_pred)**2)

def gradient_descent(x,y, learning_rate =  0.01 , epoch = 1000 ):
    m = 0 
    b = 0
    n = len(x)

    for epoch in range(epoch):
        y_pred = predict(x,m,b)
        m  =  m - learning_rate * (2/n) * np.sum((y_pred -y )*x)
        b =  b - learning_rate * (2/n) * np.sum(y_pred - y)

    return m , b 

def calculate_r2(y_actual, y_pred):
    ssr = np.sum( (y_actual- y_pred)**2)
    y_mean = np.mean(y_actual)

    sst =  np.sum((y_actual - y_mean)**2)

    r2 =  1 - (ssr/sst)

    return r2

# Dataset
x = np.array([1,2,3,4,5])
y = np.array([10,20,30,40,50])


# Train
m, b = gradient_descent(x, y)

# predictions 
y_pred = predict(x, m, b)

# Error
mse = calculate_mse(y, y_pred)
# R2 Score 
r2 = calculate_r2(y, y_pred)

print("Slope:", m)
print("Intercept:", b)
print("Predictions:", y_pred)
print("MSE:", mse)
print("R2 Score:", r2)
```

```python
# On multiple linear regression 
import numpy as np

# Y =XW+b
# prediction
def predict (X,W,b):
    return np.dot(X,W)+b

# MSE 
def mse(y, y_pred):
    return np.mean((y - y_pred) ** 2)

def r2_score(y, y_pred):

    ss_res = np.sum((y - y_pred) ** 2)

    ss_tot = np.sum((y - np.mean(y)) ** 2)

    return 1 - (ss_res / ss_tot)

def gradient_descent(X,y,learning_rate=0.0000001,
                     epochs=10000):

    m , n = X.shape

    W = np.zeros(n)

    b = 0 

    for epoch in range(epochs):
        y_pred = predict(X,W,b)

        W = W - learning_rate * (2/m) * np.dot(X.T ,(y_pred - y))
        b = b -learning_rate * (2/m) * np.sum(y_pred-y)

    return W,b

    

X = np.array([
    [1000, 2],
    [1200, 3],
    [1500, 3],
    [1800, 4]
])

y = np.array([
    200000,
    250000,
    300000,
    350000
])


W, b = gradient_descent(X, y)

print("Weights:", W)
print("Bias:", b)



y_pred = predict(X, W, b)

print("MSE:", mse(y, y_pred))

print("R2:", r2_score(y, y_pred))
```
