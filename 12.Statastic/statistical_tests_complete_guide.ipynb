## 1. Master Test-Selection Table

| Test | When to use | Data | Main question | Python |
|---|---|---|---|---|
| One-Sample Z | One mean vs claimed mean; population σ known or large-sample setting | 1 numerical variable | Is μ equal to μ₀? | `ztest()` |
| One-Sample t | One mean vs claimed mean; σ unknown | 1 numerical variable | Is μ equal to μ₀? | `ttest_1samp()` |
| Two-Sample t | Two independent means | Numerical outcome + 2 groups | Are μ₁ and μ₂ different? | `ttest_ind()` |
| Paired t | Same subjects measured twice / matched pairs | Two related numerical measurements | Did the mean change? | `ttest_rel()` |
| One-Way ANOVA | 3+ independent means, one factor | Numerical + 1 categorical factor | Are group means all equal? | `f_oneway()` |
| Two-Way ANOVA | Two categorical factors | Numerical outcome + 2 factors | Do factor A, factor B, or interaction matter? | `anova_lm()` |
| One-Sample Variance | One variance vs claimed variance | 1 numerical variable | Is σ² equal to σ₀²? | `chi2` |
| Two-Sample Variance | Compare variability of two groups | Numerical + 2 groups | Are variances/spreads different? | `levene()` |
| One-Proportion | One proportion vs claimed proportion | Binary outcome | Is p equal to p₀? | `proportions_ztest()` |
| Two-Proportion | Two independent proportions | Binary outcome + 2 groups | Are p₁ and p₂ different? | `proportions_ztest()` |
| One-Poisson Rate | One event rate vs claimed rate | Counts + exposure time | Is λ equal to λ₀? | `poisson` |
| Two-Poisson Rate | Compare two event rates | Counts + exposure time + 2 groups | Are λ₁ and λ₂ different? | `poisson` / normal approximation |
| Goodness-of-Fit | Observed frequencies vs expected distribution | Categorical/count data | Does data fit expected distribution? | `chisquare()` |
| Chi-Square Association | Association between two categorical variables | 2 categorical variables | Are the variables independent? | `chi2_contingency()` |

### Quick decision rules

- **1 mean:** Z or t
- **2 independent means:** two-sample t
- **2 related means:** paired t
- **3+ means:** ANOVA
- **1 proportion:** one-proportion test
- **2 proportions:** two-proportion test
- **1 rate:** one-Poisson rate
- **2 rates:** two-Poisson rate
- **Observed vs expected categories:** goodness-of-fit
- **Two categorical variables:** chi-square association


## 2. Common Hypothesis-Test Decision Rule

For most tests:

- If **p-value < α** → **Reject H₀**
- If **p-value ≥ α** → **Fail to reject H₀**

Common significance level:

```python
alpha = 0.05
```

### Tail selection

| Alternative hypothesis | Tail |
|---|---|
| H₁: parameter > claimed value | Right-tailed |
| H₁: parameter < claimed value | Left-tailed |
| H₁: parameter ≠ claimed value | Two-tailed |


# 3. One-Sample Z Test

### When to use

Use a one-sample Z test to compare one sample mean with a claimed population mean when the population standard deviation is known or the large-sample Z approximation is appropriate.

### Hypotheses

H₀: μ = μ₀

H₁: μ ≠ μ₀ (two-tailed example)

### Python


````python
from statsmodels.stats.weightstats import ztest

sample = [10, 11, 9, 10, 12, 11, 10]

z_stat, p_value = ztest(sample, value=10)

print("Z statistic:", z_stat)
print("P-value:", p_value)
````

### Visualization

Recommended:

- Histogram
- Boxplot

The graph describes the sample; the hypothesis test provides the formal statistical decision.


# 4. One-Sample t Test

### When to use

Use when you have one numerical sample and want to compare its mean with a claimed value, with population σ unknown.

Example: Is average delivery time different from 30 minutes?

### Python


````python
from scipy.stats import ttest_1samp

sample = [28, 32, 30, 35, 29, 31, 34, 27]

t_stat, p_value = ttest_1samp(sample, 30)

print("t statistic:", t_stat)
print("P-value:", p_value)
````

### Visualization

- Histogram
- Boxplot


# 5. Two-Sample t Test

### When to use

Compare means of two **independent** groups.

Example: Compare average production time of Machine A and Machine B.

### Hypotheses

H₀: μ₁ = μ₂

H₁: μ₁ ≠ μ₂

### Python


````python
from scipy.stats import ttest_ind

group1 = [20, 22, 19, 21, 23]
group2 = [25, 27, 24, 26, 28]

# Welch's t-test is generally preferred when equal variances
# cannot reasonably be assumed.
t_stat, p_value = ttest_ind(group1, group2, equal_var=False)

print("t statistic:", t_stat)
print("P-value:", p_value)
````

### Visualization

- Side-by-side boxplot
- Violin plot

Use `equal_var=True` for the pooled/equal-variance version when that assumption is justified.


# 6. Paired t Test

### When to use

Use when measurements are related:

- Before vs after
- Same people measured twice
- Matched pairs

### Python


````python
from scipy.stats import ttest_rel

before = [60, 55, 70, 65, 50]
after  = [65, 60, 72, 70, 58]

t_stat, p_value = ttest_rel(after, before)

print("t statistic:", t_stat)
print("P-value:", p_value)
````

### Visualization

A paired before-after plot is especially useful because it shows the change for each subject.


# 7. One-Way ANOVA

### When to use

Compare means of **3 or more independent groups** using one categorical factor.

Example: Compare test scores under Teaching Methods A, B, and C.

### Hypotheses

H₀: μ₁ = μ₂ = μ₃ = ...

H₁: At least one population mean is different.


````python
from scipy.stats import f_oneway

A = [70, 72, 68, 75, 70]
B = [75, 78, 74, 77, 76]
C = [85, 88, 90, 87, 89]

F_stat, p_value = f_oneway(A, B, C)

print("F statistic:", F_stat)
print("P-value:", p_value)
````

### Visualization

Use:

- Boxplot
- Violin plot

ANOVA tells you whether there is evidence that at least one mean differs. A post-hoc test such as Tukey HSD can identify which pairs differ.


````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_anova = pd.DataFrame({
    "Score": A + B + C,
    "Method": ["A"]*5 + ["B"]*5 + ["C"]*5
})

sns.boxplot(data=df_anova, x="Method", y="Score")
plt.title("One-Way ANOVA: Scores by Teaching Method")
plt.show()
````

# 9. One-Sample Variance Test

### When to use

Test whether a population variance equals a claimed variance.

### Formula

χ² = (n − 1)s² / σ₀²

Degrees of freedom:

df = n − 1


````python
from scipy.stats import chi2

n = 10
sample_variance = 6
claimed_variance = 4

chi_stat = (n - 1) * sample_variance / claimed_variance

# Two-tailed p-value
p_value = 2 * min(
    chi2.cdf(chi_stat, n - 1),
    chi2.sf(chi_stat, n - 1)
)

print("Chi-square statistic:", chi_stat)
print("P-value:", p_value)
````

### Visualization

- Histogram
- Boxplot

The test itself is based on the variance statistic, not on the graph.


# 10. Two-Sample Variance Test

### When to use

Compare variability between two independent groups.

A practical choice is **Levene's test**, which is more robust than Bartlett's test when normality is questionable.


````python
from scipy.stats import levene

A = [20, 22, 19, 21, 23]
B = [25, 27, 24, 26, 28]

stat, p_value = levene(A, B)

print("Levene statistic:", stat)
print("P-value:", p_value)
````

### Visualization

Use side-by-side boxplots to compare spread.


# 11. One-Proportion Test

### When to use

Compare one observed proportion with a claimed population proportion.

Example: A company claims 80% customer satisfaction.

### Python


````python
from statsmodels.stats.proportion import proportions_ztest

x = 72
n = 100
p0 = 0.80

z_stat, p_value = proportions_ztest(
    count=x,
    nobs=n,
    value=p0,
    alternative='smaller'
)

print("Z statistic:", z_stat)
print("P-value:", p_value)
````

### Alternative options

- `'smaller'` → left-tailed
- `'larger'` → right-tailed
- `'two-sided'` → two-tailed

### Visualization

Bar chart of the observed proportion.


# 12. Two-Proportion Test

### When to use

Compare proportions from two independent groups.

Example: Compare purchase rates for Campaign A and Campaign B.


````python
from statsmodels.stats.proportion import proportions_ztest

success = [80, 65]
sample_size = [100, 100]

z_stat, p_value = proportions_ztest(
    success,
    sample_size,
    alternative='larger'
)

print("Z statistic:", z_stat)
print("P-value:", p_value)
````

### Visualization

Grouped bar chart showing the two proportions.


# 13. One-Poisson Rate Test

### When to use

Test one event rate against a claimed rate.

Examples:

- Calls per hour
- Accidents per day
- Machine failures per 1,000 hours

Example: A call center claims 5 calls/hour. We observe 65 calls in 10 hours.


````python
from scipy.stats import poisson

x = 65
claimed_rate = 5
time = 10

expected_count = claimed_rate * time

# Right-tail test: P(X >= x)
p_value = poisson.sf(x - 1, expected_count)

print("Expected count under H0:", expected_count)
print("P-value:", p_value)
````

### Visualization

Use an observed-rate vs claimed-rate bar chart or rate plot.


# 14. Two-Poisson Rate Test

### When to use

Compare two independent event rates.

Example:

- Machine A: 30 failures / 100 hours
- Machine B: 20 failures / 100 hours

Estimated rates:

λ₁ = x₁ / T₁

λ₂ = x₂ / T₂

For sufficiently large counts, a normal approximation can be used.


````python
from scipy.stats import norm
import math

x1, T1 = 30, 100
x2, T2 = 20, 100

rate1 = x1 / T1
rate2 = x2 / T2

pooled_rate = (x1 + x2) / (T1 + T2)

z = (rate1 - rate2) / math.sqrt(
    pooled_rate * (1/T1 + 1/T2)
)

# Right-tailed alternative: lambda1 > lambda2
p_value = norm.sf(z)

print("Rate 1:", rate1)
print("Rate 2:", rate2)
print("Z statistic:", z)
print("P-value:", p_value)
````

### Visualization

Use a rate comparison plot or bar chart.

For small counts, consider an exact Poisson-based method rather than relying on a normal approximation.


# 15. Goodness-of-Fit Test

### When to use

Determine whether observed frequencies fit an expected distribution.

Example: A fair die should have equal expected frequencies for faces 1–6.


````python
from scipy.stats import chisquare

observed = [8, 12, 9, 11, 10, 10]
expected = [10, 10, 10, 10, 10, 10]

chi_stat, p_value = chisquare(
    observed,
    expected
)

print("Chi-square statistic:", chi_stat)
print("P-value:", p_value)
````

### Visualization

Best choice: **Observed vs Expected bar chart**.


# 16. Degree of Association — Chi-Square Test of Independence

### When to use

Use this when you have **two categorical variables** and want to determine whether they are statistically associated.

Example:

> Is product preference associated with group membership?

### Hypotheses

H₀: The two categorical variables are independent.

H₁: The two categorical variables are associated.


````python
from scipy.stats import chi2_contingency

table = [
    [40, 20],
    [25, 35]
]

chi_stat, p_value, df, expected = chi2_contingency(table)

print("Chi-square statistic:", chi_stat)
print("P-value:", p_value)
print("Degrees of freedom:", df)
print("Expected frequencies:")
print(expected)
````

### Visualization

- Stacked bar chart
- Grouped bar chart
- Heatmap

A heatmap is especially useful for larger contingency tables.

# 18. Final Exam Cheat Sheet

Ask these questions in order:

### 1. What parameter am I testing?

- Mean → Z/t/ANOVA
- Proportion → proportion test
- Variance → variance test
- Count/rate → Poisson rate
- Category frequencies → chi-square

### 2. How many groups?

- 1 group → one-sample test
- 2 independent groups → two-sample test
- 2 related measurements → paired t
- 3+ means → ANOVA

### 3. Are the variables numerical or categorical?

- Numerical outcome + categorical groups → t-test / ANOVA
- Two categorical variables → chi-square association
- One categorical variable vs expected frequencies → goodness-of-fit

### 4. What is the alternative?

- `>` → right-tail
- `<` → left-tail
- `!=` → two-tail

### 5. Final decision

```python
if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

Remember: **Rejecting H₀ means there is evidence against the null hypothesis; failing to reject H₀ does not prove H₀ is true.**
