```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
print("Original shape:", iris.frame.shape)
print("Original class names:", list(iris.target_names))
display(iris.frame.head())

df = iris.frame[iris.frame["target"].isin([1, 2])].copy()

feature_names = ["petal length (cm)", "petal width (cm)"]
X = df[feature_names]
y = (df["target"] == 2).astype(int)  # versicolor=0, virginica=1
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))


def binary_cross_entropy(y, y_pred):
    epsilon =1e-15
    y_pred = np.clip(
        y_pred,
        epsilon ,
        1-epsilon
    )
    loss = -np.mean(y*np.log(y_pred) + (1-y)*np.log(1-y_pred))


    return loss
```

```python
# y^​=σ(XW+b)
def predict_probability(X, W, b):
    z = np.dot(X, W) + b
    return sigmoid(z)
```

```python
def logistic_regression_train(
        X,
        y,
        learning_rate=0.01,
        epochs=1000):

    m ,n  =X.shape

    W = np.zeros(n)
    b = 0 
    for epoch in range(epochs):
        y_pred = predict_probability(X, W, b)

        dW = (1/m) * np.dot(
            X.T,
            (y_pred - y)
        )
        db = (1/m) * np.sum(
            y_pred - y
        )

        W -= learning_rate * dW

        b -= learning_rate * db

    return W, b

    
```

```python
def predict_class(X, W, b):

    probabilities = predict_probability(
        X, W, b
    )

    return (probabilities >= 0.5).astype(int)
```

```python
def accuracy(y_true, y_pred):

    return np.mean(
        y_true == y_pred
    )
```

```python
W, b = logistic_regression_train(
    X,
    y,
    learning_rate=0.1,
    epochs=5000
)

predictions = predict_class(
    X,
    W,
    b
)

print("Weights:", W)
print("Bias:", b)

print("Predictions:")
print(predictions)

print(
    "Accuracy:",
    accuracy(y, predictions)
)
```

```python
def evaluate_classification(y_true, y_pred):

    TP = np.sum((y_true == 1) & (y_pred == 1))

    TN = np.sum((y_true == 0) & (y_pred == 0))

    FP = np.sum((y_true == 0) & (y_pred == 1))

    FN = np.sum((y_true == 1) & (y_pred == 0))

    accuracy = (
        TP + TN
    ) / (
        TP + TN + FP + FN
    )

    precision = TP / (TP + FP + 1e-10)

    recall = TP / (TP + FN + 1e-10)

    f1 = (
        2 * precision * recall
    ) / (
        precision + recall + 1e-10
    )

    specificity = TN / (TN + FP + 1e-10)

    return {
        "TP": TP,
        "TN": TN,
        "FP": FP,
        "FN": FN,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Specificity": specificity
    }
```

```python
metrics = evaluate_classification(
    y,
    predictions
)

for key, value in metrics.items():
    print(key, ":", value)
```
