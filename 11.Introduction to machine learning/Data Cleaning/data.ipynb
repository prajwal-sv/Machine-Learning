```python
df["Age"] = df["Age"].astype(str).str.replace("yrs", "", regex=False).str.strip()
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python

df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
for col in ["Name", "Sex", "Embarked"]:
    df[col] = df[col].str.strip()
```

```python
import numpy as np 
df.loc[df["Age"] > 100, "Age"] = np.nan
df.loc[df["Age"] < 0, "Age"] = np.nan

df["Age"] = df["Age"].fillna(df["Age"].median())

```

```python
df["Fare"] = df["Fare"].astype(str).str.replace("£", "", regex=False).str.strip()
```

```python
df["Fare"] = pd.to_numeric(df["Fare"], errors="coerce")
df.loc[(df["Fare"] < 0) | (df["Fare"] > 600), "Fare"] = np.nan
```

```python
df.loc[df["SibSp"] < 0, "SibSp"] = np.nan
df.loc[df["Parch"] < 0, "Parch"] = np.nan
```
