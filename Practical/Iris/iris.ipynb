```python
import pandas as pd 

df = pd.read_csv('Dirty_Iris_Clustering.csv')
```

```python
df.info()
```

```python

# ==========================================
# 1. IMPORT LIBRARIES AND LOAD DATASET
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

# Load dataset


# Display dataset
print(df.head())

# Dataset information
print(df.shape)
print(df.info())
print(df.describe())

# Missing values
print(df.isnull().sum())

# Duplicate values
print("Duplicates:", df.duplicated().sum())


# ==========================================
# 2. EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================

# Convert petal length into numeric
df["petal_length_cm"] = pd.to_numeric(
    df["petal_length_cm"], errors="coerce"
)

# Distribution plots
df[["sepal_length_cm", "sepal_width_cm",
    "petal_length_cm", "petal_width_cm"]].hist(
    figsize=(10, 7)
)

plt.tight_layout()
plt.show()

# Boxplot
sns.boxplot(data=df[[
    "sepal_length_cm", "sepal_width_cm",
    "petal_length_cm", "petal_width_cm"
]])
plt.title("Outlier Detection")
plt.show()

# Correlation heatmap
sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Feature Correlation")
plt.show()


# ==========================================
# 3. DATA CLEANING
# ==========================================

# Remove duplicates
df = df.drop_duplicates()

# Drop unnecessary columns
df = df.drop(columns=[
    "plant_record_id",
    "collector_notes",
    "soil_ph_test"
])

# Encode categorical column
df = pd.get_dummies(
    df,
    columns=["collection_site"],
    dtype=int
)

# Select clustering features
X = df[[
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]].copy()

# Handle missing values using median
X = X.fillna(X.median())

print("Missing values after cleaning:")
print(X.isnull().sum())


# ==========================================
# 4. FEATURE SCALING
# ==========================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaled data:")
print(X_scaled[:5])


# ==========================================
# 5. BASELINE MODEL 1 - KMEANS
# ==========================================

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df["KMeans_Cluster"] = kmeans.fit_predict(X_scaled)

print("KMeans Cluster Sizes:")
print(df["KMeans_Cluster"].value_counts())

# Cluster visualization
plt.scatter(
    X_scaled[:, 2],
    X_scaled[:, 3],
    c=df["KMeans_Cluster"]
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("KMeans Clustering")
plt.show()


# ==========================================
# 6. BASELINE MODEL 2 - AGGLOMERATIVE
# ==========================================

agg = AgglomerativeClustering(n_clusters=3)

df["Agg_Cluster"] = agg.fit_predict(X_scaled)

print("Agglomerative Cluster Sizes:")
print(df["Agg_Cluster"].value_counts())

# Compare both algorithms
print("\nKMeans:")
print(df["KMeans_Cluster"].value_counts())

print("\nAgglomerative:")
print(df["Agg_Cluster"].value_counts())


# ==========================================
# 7. ELBOW METHOD
# ==========================================

wcss = []

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)

    wcss.append(model.inertia_)

plt.plot(range(2, 11), wcss, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()


# ==========================================
# 8. SILHOUETTE SCORE
# ==========================================

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    print("K =", k, "Silhouette Score =", round(score, 3))


# ==========================================
# 9. FINAL KMEANS MODEL
# ==========================================

# Select k after checking elbow and silhouette results
best_k = 3

final_model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df["Final_Cluster"] = final_model.fit_predict(X_scaled)

print("Final Cluster Sizes:")
print(df["Final_Cluster"].value_counts())

# Cluster profiling
print("\nCluster Profile:")
print(df.groupby("Final_Cluster")[[
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]].mean())

# Final silhouette score
print(
    "Final Silhouette Score:",
    silhouette_score(X_scaled, df["Final_Cluster"])
)


# ==========================================
# 10. PCA DIMENSIONALITY REDUCTION
# ==========================================

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Explained Variance:")
print(pca.explained_variance_ratio_)

print(
    "Total Variance Explained:",
    sum(pca.explained_variance_ratio_) * 100
)


# ==========================================
# 11. PCA VISUALIZATION
# ==========================================

plt.figure(figsize=(8, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df["Final_Cluster"],
    cmap="viridis"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Clustering Visualization")
plt.colorbar(label="Cluster")
plt.show()


# ==========================================
# 12. FINAL CLUSTER OBSERVATIONS
# ==========================================

print("\nFINAL CLUSTER DISTRIBUTION")
print(df["Final_Cluster"].value_counts())

print("\nFINAL CLUSTER MEAN")
print(df.groupby("Final_Cluster")[[
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]].mean())

print("""
OBSERVATIONS:

1. KMeans groups similar flower measurements.
2. Agglomerative clustering provides another clustering result.
3. The elbow method helps identify the suitable number of clusters.
4. Silhouette score measures cluster separation.
5. PCA reduces four features into two components.
6. The PCA plot shows the distribution of clusters.
7. Cluster profiling explains the differences between groups.

Clustering is unsupervised, so train-test split
and GridSearchCV are not required.
""")

print("CLUSTERING WITH PCA COMPLETED!")
```

```python

# ==========================================
# 9. HYPERPARAMETER TUNING
# ==========================================

results = []

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        init="k-means++",
        n_init=10,
        max_iter=300,
        random_state=42
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    results.append({
        "K": k,
        "Inertia": model.inertia_,
        "Silhouette Score": score
    })

# Display results
results_df = pd.DataFrame(results)

print(results_df)

# Select best K using silhouette score
best_k = results_df.loc[
    results_df["Silhouette Score"].idxmax(), "K"
]

print("\nBest K:", best_k)

# Train final tuned model
tuned_model = KMeans(
    n_clusters=int(best_k),
    init="k-means++",
    n_init=10,
    max_iter=300,
    random_state=42
)

df["Final_Cluster"] = tuned_model.fit_predict(X_scaled)

# Final evaluation
print("\nFinal Silhouette Score:")
print(silhouette_score(X_scaled, df["Final_Cluster"]))

print("\nFinal Cluster Sizes:")
print(df["Final_Cluster"].value_counts().sort_index())

# Plot tuning results
plt.plot(
    results_df["K"],
    results_df["Silhouette Score"],
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("KMeans Hyperparameter Tuning")
plt.show()
```
